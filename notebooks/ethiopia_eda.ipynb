```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

📊 Correlation Analysis (Ethiopia)
Temperature (T2M) is strongly negatively correlated with relative humidity (RH2M), meaning higher temperatures are associated with lower humidity levels.
Precipitation (PRECTOTCORR) shows weak correlation with temperature, suggesting rainfall is not directly driven by daily temperature changes.
Wind speed (WS2M) has very weak correlation with other variables, indicating it behaves independently in this dataset.
Maximum temperature (T2M_MAX) and mean temperature (T2M) are strongly positively correlated, which is expected since they measure related thermal conditions.
